```python
#정수 하나를 입력받아 그 수의 구구단을 7 x 1 = 7 형식으로 출력하세요
gugudan = int(input('단을 입력하세요: '))

for i in range(1, 10):
    print(f'{gugudan} x {i} = {gugudan * i}')
```

```python
# 2단부터 9단까지 구구단을 7*1=7 형식을 출력하세요, 단이 끝날 때 마다 점선을 넣고...
for i in range (2,10):
  for j in range (1,10):
    print(f'{i} x {j} = {i*j}')

  print('-'*10)
```

```python
#정수 n을 입력받아 n의 약수의 개수를 출력하고, 약수가 2개이면 "소수입니다."를 함께 출력하세요.
n = int(input('정수를 입력하세요: '))
count = 0

for a in range(1, n + 1):
    if n % a == 0:  # n을 a로 나누어 떨어지면 (약수이면)
        count += 1  # count = count + 1

print(f'약수의 개수: {count}개')

if count == 2:
    print("소수입니다.")
```

```python
#while 문으로 1에서 100까지 더해보기
a=1
sum=0 #변수 초기화가 필요
while a<101:
  sum=sum+a
  a=a+1

print(sum)
```

```python
sum = 0
i = 1

while True:
    if i > 100:
        break
    sum += i
    i += 1

print(sum)
```

```python
#양의 정수를 계속 입력받아 합계를 구하다가 0을 입력하면 반복을 끝내고 합계를 출력하세요
sum = 0
while True:
    n = int(input('양의 정수를 입력하세요: '))
    if n == 0:
        break
    sum += n
print(sum)
```

```python
#1~20까지 중 짝수만 담은 리스트를 내포로 만들어 출력하시오
evens = []
for i in range(1, 21):
    if  i%2 == 0:
        evens.append(i)
print(evens)
evens.pop(7) #8번째, 16을 지움 -> 반환값 가능
print(evens)
del evens[5]
print(evens)
print(evens[3])
print(evens[-1])
```

```python
#sum 사용 여부에 따른 차이
scores = [85, 95, 70, 75, 99, 93, 88]

print('최댓값:', max(scores))
print('최솟값:', min(scores))
print('합:', sum(scores))
print('평균:', sum(scores)/len(scores))

#sum을 이용하여 합계 구하기 -> 여기에서의 sum은 함수다
sum(scores)

print(max(scores))
```

```python
#반복문을 활용하여 더해보자
yy = 0
scores = [85, 95, 70, 75, 99, 93, 88]
for i in scores:
  yy += i
print(yy)
```
