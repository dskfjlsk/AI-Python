```python
import random

a = random.randint(1, 100)
b = random.randint(1, 100)
print('교환 전:', a, b)

if a < b:
    a, b = b, a        # 2장에서 배운 값 교환

print('큰 수:', a, '작은 수:', b)
```

```python
height1 = float(input('첫 번째 학생의 키(cm): '))
height2 = float(input('두 번째 학생의 키(cm): '))

if height1 > height2:
    diff = height1 - height2
else:
    diff = height2 - height1

print(f'두 학생의 키 차이는 {diff:.1f}cm입니다.')
```

```python
# 1: 가위, 2: 바위, 3: 보
com = random.randint(1, 3)
user = int(input('가위(1), 바위(2), 보(3) 중 선택: '))

if com == 1:
    print('컴퓨터: 가위')
elif com == 2:
    print('컴퓨터: 바위')
else:
    print('컴퓨터: 보')

if user == com:
    print('비겼습니다.')
elif user == 1 and com == 3:
    print('이겼습니다!')        # 가위 vs 보
elif user == 2 and com == 1:
    print('이겼습니다!')        # 바위 vs 가위
elif user == 3 and com == 2:
    print('이겼습니다!')        # 보 vs 바위
else:
    print('졌습니다.')
```

```python
import random

com = random.randint(1, 100)

print("--- 숫자 맞추기 게임 (1~100) ---")

while True:
    user = int(input('1에서 100 사이 숫자를 입력하세요: '))

    if user < com:
        print('정답보다 작습니다. (Up)\n')
    elif user > com:
        print('정답보다 큽니다. (Down)\n')
    else:
        print('정답입니다! 🎉')
        break 
```

```python
#asmd 함수정의
def asmd(x, y, op):
    if op == '+':
        print(x + y)
    elif op == '-':
        print(x - y)
    elif op == '*':
        print(x * y)
    elif op == '/':
        if y == 0:
            print("0으로 나눌 수 없습니다.")
        else:
            print(x / y)
    else:
        print("올바른 연산자가 아닙니다.")
```

```python
#교수님 ver.
def asmd(x, y, op):
    if op == '+':
        print(f'{x} + {y} = {x + y}')
    elif op == '-':
        print(f'{x} - {y} = {x - y}')
    elif op == '*':
        print(f'{x} * {y} = {x * y}')
    elif op == '/':
        if y == 0:
            print('0으로 나눌 수 없습니다.')
        else:
            print(f'{x} / {y} = {x / y}')
    else:
        print('올바른 연산자가 아닙니다.')
```

```python
x=int(input('첫 번째 수: '))
y=int(input('두 번째 수: '))
op=input('연산자를 입력하시오')
asmd(x,y,op)
```

```python
#5번 반복문

import random

com = random.randint(1, 100)

for i in range(5):
    user = int(input('숫자를 입력하세요'))
    # print(com)
    if user == com:
        print('정답입니다')
        break  # 종료시킴
    elif user > com:
        print('정답보다 큽니다')
    else:
        print('정답보다 작습니다')
```

```python
# 무한 반복문 + 몇 번만에 맞추었는가?
import random

com = random.randint(1, 100)
count = 0 

while True:
    user = int(input('숫자를 입력하세요'))
    count += 1
    
    if user == com:
        print(f'정답입니다! ({count}번 만에 맞힘)')
        break # 종료시킴
    elif user > com:
        print('정답보다 큽니다')
    else:
        print('정답보다 작습니다')
```
